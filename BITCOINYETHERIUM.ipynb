TABLA DE MÉTRICAS FINANCIERAS

```python
# ── Tabla de métricas + Drawdown histórico ──────────────────────────

filas = []

drawdown = pd.DataFrame()

for cripto, nombre in zip(CRIPTOS, NOMBRES):

    # ── Precios ─────────────────────────────────────────────────────

    precios = datos[cripto].dropna()

    # ── Rendimientos ────────────────────────────────────────────────

    rend_log = np.log(precios / precios.shift(1)).dropna()

    rend_simp = precios.pct_change().dropna()

    # ── Retorno total (%) ───────────────────────────────────────────

    retorno_total = (
        (precios.iloc[-1] / precios.iloc[0]) - 1
    ) * 100

    # ── Volatilidad anualizada ──────────────────────────────────────

    volatilidad = rend_log.std() * np.sqrt(365)

    # ── Curva de riqueza ────────────────────────────────────────────

    riqueza = (1 + rend_simp).cumprod()

    # ── Max Drawdown (%) ────────────────────────────────────────────

    dd = (riqueza / riqueza.cummax()) - 1

    max_drawdown = dd.min()

    drawdown[cripto] = dd

    # ── % días negativos ────────────────────────────────────────────

    pct_negativos = (
        (rend_simp < 0).mean()
    ) * 100

    filas.append({

        "Criptomoneda": nombre,

        "Retorno total (%)":
            round(retorno_total, 1),

        "Volatilidad anual (%)":
            round(volatilidad * 100, 1),

        "Max Drawdown (%)":
            round(max_drawdown * 100, 1),

        "% Días negativos":
            round(pct_negativos, 1),
    })

# ── Tabla final ─────────────────────────────────────────────────────

tabla_metricas = pd.DataFrame(filas).set_index("Criptomoneda")

# ── Mostrar tabla premium ───────────────────────────────────────────

display(

    tabla_metricas.style
    .background_gradient(
        subset=["Retorno total (%)"],
        cmap="RdYlGn"
    )
    .background_gradient(
        subset=["Volatilidad anual (%)"],
        cmap="YlOrRd"
    )
    .background_gradient(
        subset=["Max Drawdown (%)"],
        cmap="RdYlGn"
    )
    .background_gradient(
        subset=["% Días negativos"],
        cmap="YlOrRd"
    )
    .format("{:.1f}")

)

# ── Gráfica Drawdown Histórico ──────────────────────────────────────

fig = go.Figure()

for i, cripto in enumerate(CRIPTOS):

    fig.add_trace(
        go.Scatter(

            x=drawdown.index,

            y=drawdown[cripto] * 100,

            mode='lines',

            name=NOMBRES[i],

            line=dict(
                color=COLORES[i],
                width=3
            )
        )
    )

fig.update_layout(

    title=dict(
        text="📉 Drawdown Histórico — Caída desde el Pico",
        x=0.5,
        font=dict(size=24)
    ),

    template="plotly_dark",

    height=650,

    paper_bgcolor="#0B1020",
    plot_bgcolor="#050816",

    font=dict(color="white"),

    hovermode="x unified"
)

fig.update_xaxes(
    title="Fecha",
    tickformat="%m/%Y"
)

fig.update_yaxes(
    title="Drawdown (%)"
)

fig.show()
```

Estrategia de Tendencia Adaptativa — Vector Capital Digital

Concepto

La estrategia del fondo utiliza medias móviles para identificar tendencias del mercado.

Cuando el precio de Bitcoin se mantiene por encima de la media móvil:
👉 el fondo participa en el mercado.
Cuando el precio cae por debajo:
👉 el fondo reduce exposición y protege capital.

La estrategia busca:

aprovechar tendencias favorables,
reducir exposición en escenarios de corrección,
y mejorar el control del riesgo frente a mantener la inversión pasivamente.


```python
# ── Celda 3.1 — Estrategia Vector Capital Digital ───────────────────

VENTANA_SMA = 30

# ── Bitcoin para estrategia ─────────────────────────────────────────

precios_bt = datos["BTC-USD"].dropna()

# ── Media móvil ─────────────────────────────────────────────────────

sma = precios_bt.rolling(
    window=VENTANA_SMA
).mean()

# ── Señal de inversión ──────────────────────────────────────────────

señal = (precios_bt > sma).astype(int)

# ── Subplots ────────────────────────────────────────────────────────

fig = make_subplots(

    rows=2,
    cols=1,

    shared_xaxes=True,

    row_heights=[0.75, 0.25],

    vertical_spacing=0.08,

    subplot_titles=[

        f"📈 Bitcoin vs Media Móvil ({VENTANA_SMA} días)",

        "🟢 Posición del Fondo"
    ]
)

# ── Precio Bitcoin ──────────────────────────────────────────────────

fig.add_trace(

    go.Scatter(

        x=precios_bt.index,

        y=precios_bt.values,

        name="Bitcoin",

        line=dict(
            color="#F7931A",
            width=2.5
        )
    ),

    row=1,
    col=1
)

# ── Media móvil ─────────────────────────────────────────────────────

fig.add_trace(

    go.Scatter(

        x=sma.index,

        y=sma.values,

        name=f"SMA {VENTANA_SMA}",

        line=dict(
            color="#D4A437",
            width=2,
            dash="dot"
        )
    ),

    row=1,
    col=1
)

# ── Señal ───────────────────────────────────────────────────────────

fig.add_trace(

    go.Scatter(

        x=señal.index,

        y=señal.values.flatten(),

        fill='tozeroy',

        name="Posición",

        line=dict(
            color="#2ECC71",
            width=2
        ),

        fillcolor="rgba(46,204,113,0.25)"
    ),

    row=2,
    col=1
)

# ── Diseño Premium ──────────────────────────────────────────────────

fig.update_layout(

    title=dict(

        text="📊 Estrategia de Tendencia Adaptativa — Vector Capital Digital",

        x=0.5,

        y=0.96,

        font=dict(
            size=24
        )
    ),

    template="plotly_dark",

    height=650,

    hovermode="x unified",

    paper_bgcolor="#0B1020",

    plot_bgcolor="#050816",

    font=dict(
        color="white",
        family="Arial"
    ),

    legend=dict(

        orientation="h",

        yanchor="bottom",

        y=1.01,

        xanchor="center",

        x=0.5
    ),

    margin=dict(
        l=40,
        r=40,
        t=80,
        b=40
    )
)

# ── Subtítulos más abajo ────────────────────────────────────────────

fig.update_annotations(

    font=dict(size=15),

    yshift=-8
)

# ── Eje X ───────────────────────────────────────────────────────────

fig.update_xaxes(

    title="Fecha",

    tickformat="%m/%Y",

    showgrid=True,

    gridcolor="rgba(100,100,255,0.10)"
)

# ── Eje Y ───────────────────────────────────────────────────────────

fig.update_yaxes(

    showgrid=True,

    gridcolor="rgba(100,100,255,0.10)"
)

fig.show()

# ── Estadística estrategia ──────────────────────────────────────────

dias_inv = int(señal.sum())

print(
    f"⏱️ Días invertido: {dias_inv} de {len(señal.dropna())} "
    f"({dias_inv/len(señal.dropna())*100:.1f}%)"
)
```

BACKTESTING

```python
señal.shift(1)
```

Concepto: evitamos el lookahead bias aplicando señal.shift(1), ya que la señal del día solo se conoce al cierre del mercado. La estrategia actúa al día siguiente utilizando información disponible de manera realista.

```python
# ── Celda 3.2 — Backtesting Estrategia SMA ──────────────────────────

VENTANA_SMA = 30

CAPITAL_INICIAL = 10_000

# ── Bitcoin ─────────────────────────────────────────────────────────

precios_bt = datos["BTC-USD"].dropna()

# ── Rendimientos diarios ────────────────────────────────────────────

retornos_bt = precios_bt.pct_change().dropna()

# ── Media móvil ─────────────────────────────────────────────────────

sma = precios_bt.rolling(
    window=VENTANA_SMA
).mean()

# ── Señal ───────────────────────────────────────────────────────────

señal = (
    precios_bt > sma
).astype(int)

# ── Evitar Lookahead Bias ───────────────────────────────────────────

señal = señal.shift(1).fillna(0)

# ── Rendimiento estrategia ──────────────────────────────────────────

retorno_estrategia = retornos_bt * señal.loc[retornos_bt.index]

# ── Curvas acumuladas ───────────────────────────────────────────────

capital_estrategia = (
    (1 + retorno_estrategia)
    .cumprod()
) * CAPITAL_INICIAL

capital_buyhold = (
    (1 + retornos_bt)
    .cumprod()
) * CAPITAL_INICIAL

# ── Sharpe Ratio ────────────────────────────────────────────────────

sharpe_estrategia = (
    retorno_estrategia.mean() /
    retorno_estrategia.std()
) * np.sqrt(365)

sharpe_buyhold = (
    retornos_bt.mean() /
    retornos_bt.std()
) * np.sqrt(365)

# ── Gráfica Premium ─────────────────────────────────────────────────

fig = go.Figure()

# ── Estrategia ──────────────────────────────────────────────────────

fig.add_trace(

    go.Scatter(

        x=capital_estrategia.index,

        y=capital_estrategia.values,

        mode='lines',

        name=f"Estrategia SMA {VENTANA_SMA}d",

        line=dict(
            color="#2ECC71",
            width=3
        )
    )
)

# ── Buy & Hold ──────────────────────────────────────────────────────

fig.add_trace(

    go.Scatter(

        x=capital_buyhold.index,

        y=capital_buyhold.values,

        mode='lines',

        name="Buy & Hold",

        line=dict(
            color="#F7931A",
            width=3
        )
    )
)

# ── Línea capital inicial ───────────────────────────────────────────

fig.add_hline(

    y=CAPITAL_INICIAL,

    line_dash="dash",

    line_color="gray",

    annotation_text="Capital inicial"
)

# ── Diseño ──────────────────────────────────────────────────────────

fig.update_layout(

    title=dict(

        text=f"📊 Backtesting — Bitcoin vs SMA {VENTANA_SMA}d",

        x=0.5,

        font=dict(size=24)
    ),

    template="plotly_dark",

    height=650,

    hovermode="x unified",

    paper_bgcolor="#0B1020",

    plot_bgcolor="#050816",

    font=dict(
        color="white"
    ),

    legend=dict(

        orientation="h",

        yanchor="bottom",

        y=1.01,

        xanchor="center",

        x=0.5
    )
)

# ── Ejes ────────────────────────────────────────────────────────────

fig.update_xaxes(

    title="Fecha",

    tickformat="%m/%Y",

    showgrid=True,

    gridcolor="rgba(100,100,255,0.10)"
)

fig.update_yaxes(

    title="Valor del Portafolio (USD)",

    showgrid=True,

    gridcolor="rgba(100,100,255,0.10)"
)

fig.show()

# ── Resultados ──────────────────────────────────────────────────────

print(f"💰 Capital final estrategia: ${capital_estrategia.iloc[-1]:,.0f}")

print(f"💰 Capital final Buy & Hold: ${capital_buyhold.iloc[-1]:,.0f}")

print(f"📈 Sharpe estrategia: {sharpe_estrategia:.2f}")

print(f"📈 Sharpe Buy & Hold: {sharpe_buyhold:.2f}")
```

```python
%%writefile app_fondo.py

# ════════════════════════════════════════════════════════════════════════
# VECTOR CAPITAL DIGITAL
# VII Concurso de Analítica Financiera — ITM 2026
# ════════════════════════════════════════════════════════════════════════

# ── Librerías ────────────────────────────────────────────────────────

import streamlit as st
import yfinance as yf
import pandas as pd
import numpy as np

import plotly.graph_objects as go
import plotly.express as px

from plotly.subplots import make_subplots

from datetime import date, timedelta

# ════════════════════════════════════════════════════════════════════════
# CONFIGURACIÓN GENERAL
# ════════════════════════════════════════════════════════════════════════

st.set_page_config(

    page_title="VECTOR CAPITAL DIGITAL",

    page_icon="₿",

    layout="wide"
)

# ════════════════════════════════════════════════════════════════════════
# IDENTIDAD DEL FONDO
# ════════════════════════════════════════════════════════════════════════

NOMBRE_FONDO = "VECTOR CAPITAL DIGITAL"

TESIS = (
    "Convertimos volatilidad y datos del mercado "
    "en ventajas estratégicas de inversión inteligente."
)

COLOR_PRIMARIO = "#071B3B"

COLOR_ACENTO = "#D4A437"

CAPITAL_INICIAL = 10_000

# ════════════════════════════════════════════════════════════════════════
# CRIPTOMONEDAS
# ════════════════════════════════════════════════════════════════════════

CRIPTOS_DISP = {

    "BTC-USD": ("Bitcoin", "#F7931A"),

    "ETH-USD": ("Ethereum", "#627EEA"),

    "SOL-USD": ("Solana", "#00FFA3"),

    "ADA-USD": ("Cardano", "#3468D1"),

    "LTC-USD": ("Litecoin", "#B8B8B8"),
}

# ════════════════════════════════════════════════════════════════════════
# SIDEBAR
# ════════════════════════════════════════════════════════════════════════

with st.sidebar:

    st.markdown("## ⚙️ Configuración")

    st.markdown("---")

    criptos_sel = st.multiselect(

        "🪙 Criptomonedas",

        options=list(CRIPTOS_DISP.keys()),

        default=[
            "BTC-USD",
            "ETH-USD",
            "SOL-USD"
        ],

        format_func=lambda x: CRIPTOS_DISP[x][0]
    )

    col1, col2 = st.columns(2)

    with col1:

        fecha_ini = st.date_input(

            "Inicio",

            value=date.today() - timedelta(days=730)
        )

    with col2:

        fecha_fin = st.date_input(

            "Fin",

            value=date.today()
        )

    frecuencia = st.selectbox(

        "Frecuencia",

        ["Diaria", "Semanal", "Mensual"]
    )

    FREQ_MAP = {

        "Diaria": "D",

        "Semanal": "W",

        "Mensual": "ME"
    }

    capital = st.number_input(

        "💵 Capital inicial",

        min_value=1000,

        max_value=1_000_000,

        value=10_000,

        step=1000
    )

    st.markdown("---")

    ventana_sma = st.slider(

        "🎛️ Ventana SMA",

        5,

        100,

        30
    )

    st.markdown("---")

    perfil = st.selectbox(

        "Perfil de riesgo",

        ["Conservador", "Moderado", "Agresivo"]
    )

    st.markdown("---")

    st.info(

        f"**{NOMBRE_FONDO}**\n\n"

        f"{TESIS}\n\n"

        f"Perfil seleccionado: {perfil}"
    )

# ════════════════════════════════════════════════════════════════════════
# HEADER
# ════════════════════════════════════════════════════════════════════════

st.markdown(f"# ₿ {NOMBRE_FONDO}")

st.markdown(
    "### VII Concurso de Analítica Financiera — ITM 2026"
)

st.markdown("---")

# ════════════════════════════════════════════════════════════════════════
# VALIDACIÓN
# ════════════════════════════════════════════════════════════════════════

if not criptos_sel:

    st.warning(
        "Selecciona al menos una criptomoneda."
    )

    st.stop()

# ════════════════════════════════════════════════════════════════════════
# DESCARGA DE DATOS
# ════════════════════════════════════════════════════════════════════════

@st.cache_data(ttl=3600)

def cargar_datos(criptos, ini, fin):

    df = yf.download(

        criptos,

        start=ini,

        end=fin,

        auto_adjust=True
    )

    return df

with st.spinner("⏳ Descargando datos..."):

    datos = cargar_datos(

        criptos_sel,

        fecha_ini,

        fecha_fin
    )

if datos.empty:

    st.error("No se pudieron descargar datos.")

    st.stop()

freq = FREQ_MAP[frecuencia]

if freq != "D":

    datos = datos.resample(freq).last()

# ════════════════════════════════════════════════════════════════════════
# KPIs
# ════════════════════════════════════════════════════════════════════════

st.markdown("## 📊 Resumen del periodo")

cols = st.columns(len(criptos_sel))

for i, cripto in enumerate(criptos_sel):

    nombre, color = CRIPTOS_DISP[cripto]

    p = (
        datos["Close"][cripto]
        if len(criptos_sel) > 1
        else datos["Close"]
    ).dropna()

    r = np.log(
        p / p.shift(1)
    ).dropna()

    retorno = (
        (p.iloc[-1] / p.iloc[0]) - 1
    ) * 100

    volatilidad = (
        r.std() * np.sqrt(365)
    ) * 100

    delta = (
        (p.iloc[-1] / p.iloc[-2]) - 1
    ) * 100

    with cols[i]:

        st.metric(

            nombre,

            f"${p.iloc[-1]:,.2f}",

            f"{delta:+.2f}%"
        )

        st.caption(

            f"Retorno: {retorno:+.1f}% | "
            f"Volatilidad: {volatilidad:.1f}%"
        )

st.markdown("---")

# ════════════════════════════════════════════════════════════════════════
# PRECIOS HISTÓRICOS
# ════════════════════════════════════════════════════════════════════════

st.markdown("## 📈 Precios históricos")

fig_precios = go.Figure()

for cripto in criptos_sel:

    nombre, color = CRIPTOS_DISP[cripto]

    p = (
        datos["Close"][cripto]
        if len(criptos_sel) > 1
        else datos["Close"]
    ).dropna()

    fig_precios.add_trace(

        go.Scatter(

            x=p.index,

            y=p.values,

            mode="lines",

            name=nombre,

            line=dict(
                color=color,
                width=3
            )
        )
    )

fig_precios.update_layout(

    hovermode="x unified",

    height=500,

    xaxis_title="Fecha",

    yaxis_title="Precio USD",

    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02
    )
)

st.plotly_chart(

    fig_precios,

    use_container_width=True
)

# ════════════════════════════════════════════════════════════════════════
# RENDIMIENTO ACUMULADO
# ════════════════════════════════════════════════════════════════════════

st.markdown("## 💰 Rendimiento acumulado")

fig_acum = go.Figure()

for cripto in criptos_sel:

    nombre, color = CRIPTOS_DISP[cripto]

    p = (
        datos["Close"][cripto]
        if len(criptos_sel) > 1
        else datos["Close"]
    ).dropna()

    rend = np.log(
        p / p.shift(1)
    ).dropna()

    valor = capital * np.exp(
        rend.cumsum()
    )

    fig_acum.add_trace(

        go.Scatter(

            x=valor.index,

            y=valor.values,

            mode="lines",

            name=nombre,

            line=dict(
                color=color,
                width=3
            )
        )
    )

fig_acum.add_hline(

    y=capital,

    line_dash="dot",

    line_color="gray"
)

fig_acum.update_layout(

    hovermode="x unified",

    height=500,

    xaxis_title="Fecha",

    yaxis_title="Valor USD"
)

st.plotly_chart(

    fig_acum,

    use_container_width=True
)

# ════════════════════════════════════════════════════════════════════════
# MÉTRICAS
# ════════════════════════════════════════════════════════════════════════

st.markdown("## ⚠️ Métricas de riesgo")

filas = []

for cripto in criptos_sel:

    nombre, _ = CRIPTOS_DISP[cripto]

    p = (
        datos["Close"][cripto]
        if len(criptos_sel) > 1
        else datos["Close"]
    ).dropna()

    r_log = np.log(
        p / p.shift(1)
    ).dropna()

    r_simp = p.pct_change().dropna()

    retorno = (
        (p.iloc[-1] / p.iloc[0]) - 1
    ) * 100

    vol = (
        r_log.std() * np.sqrt(365)
    ) * 100

    riqueza = (
        1 + r_simp
    ).cumprod()

    drawdown = (
        riqueza / riqueza.cummax() - 1
    )

    max_dd = drawdown.min() * 100

    pct_neg = (
        (r_simp < 0).mean()
    ) * 100

    filas.append({

        "Cripto": nombre,

        "Retorno (%)":
            round(retorno, 1),

        "Volatilidad (%)":
            round(vol, 1),

        "Max Drawdown (%)":
            round(max_dd, 1),

        "% Días Negativos":
            round(pct_neg, 1)
    })

df_metricas = pd.DataFrame(filas)

st.dataframe(

    df_metricas,

    use_container_width=True
)

# ════════════════════════════════════════════════════════════════════════
# BACKTESTING
# ════════════════════════════════════════════════════════════════════════

st.markdown("## 🎯 Estrategia y Backtesting")

cripto_bt = st.selectbox(

    "Cripto para estrategia",

    criptos_sel,

    format_func=lambda x: CRIPTOS_DISP[x][0]
)

datos_bt = yf.download(

    cripto_bt,

    start=fecha_ini,

    end=fecha_fin,

    auto_adjust=True
)

p_bt = datos_bt["Close"].squeeze().dropna()

r_bt = np.log(
    p_bt / p_bt.shift(1)
).dropna()

sma_bt = p_bt.rolling(
    ventana_sma
).mean()

senal_bt = (
    p_bt > sma_bt
).astype(int)

rend_est = (
    r_bt *
    senal_bt.shift(1).fillna(0)
)

valor_est = capital * np.exp(
    rend_est.cumsum()
)

valor_bh = capital * np.exp(
    r_bt.cumsum()
)

fig_bt = go.Figure()

fig_bt.add_trace(

    go.Scatter(

        x=valor_est.index,

        y=valor_est.values,

        mode="lines",

        name="Estrategia SMA",

        line=dict(
            color="#2ECC71",
            width=3
        )
    )
)

fig_bt.add_trace(

    go.Scatter(

        x=valor_bh.index,

        y=valor_bh.values,

        mode="lines",

        name="Buy & Hold",

        line=dict(
            color="#F7931A",
            width=3,
            dash="dash"
        )
    )
)

fig_bt.update_layout(

    height=500,

    hovermode="x unified",

    xaxis_title="Fecha",

    yaxis_title="Valor del Portafolio"
)

st.plotly_chart(

    fig_bt,

    use_container_width=True
)

# ════════════════════════════════════════════════════════════════════════
# CORRELACIÓN
# ════════════════════════════════════════════════════════════════════════

st.markdown("## 🔥 Correlación de Rendimientos")

rend_todos = pd.DataFrame()

for c in criptos_sel:

    p = (
        datos["Close"][c]
        if len(criptos_sel) > 1
        else datos["Close"]
    ).dropna()

    rend_todos[
        CRIPTOS_DISP[c][0]
    ] = np.log(
        p / p.shift(1)
    )

if len(criptos_sel) > 1:

    corr = rend_todos.dropna().corr()

    fig_corr = px.imshow(

        corr,

        text_auto=".2f",

        color_continuous_scale="RdYlGn",

        zmin=0,

        zmax=1
    )

    st.plotly_chart(

        fig_corr,

        use_container_width=True
    )

# ════════════════════════════════════════════════════════════════════════
# DISCLAIMER
# ════════════════════════════════════════════════════════════════════════

st.markdown("---")

with st.expander("📋 Uso de Inteligencia Artificial"):

    st.markdown("""

| Herramienta | Uso | Validación |
|---|---|---|
| ChatGPT 5.5 | Código y visualizaciones | Estrategia y análisis financiero |
| Yahoo Finance | Datos financieros | Verificación de mercado |
| Python y Streamlit | Construcción app | Funcionamiento técnico |

> El equipo comprende y valida cada análisis presentado.
""")

st.caption(
    "VII Concurso de Analítica Financiera · ITM · 2026"
)
```
